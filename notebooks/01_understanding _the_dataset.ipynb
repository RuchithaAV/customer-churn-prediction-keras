## Phase 1: Column Decisions

**Drop (identifiers / no signal):**
- CustomerID — unique identifier, no predictive value
- Count — constant column, no signal
- Country — single value across dataset, no signal
- State — single value across dataset, no signal

**Drop (redundant geography):**
- Lat Long, Latitude, Longitude, Zip Code — redundant with each other; 
  city-level signal will be captured via City (bucketed) instead

**Defer to EDA / preprocessing:**
- City — high cardinality (1,129 unique values); will bucket top-N + "Other" 
  in Phase 3 if EDA shows meaningful churn-rate variation by city

**Drop (target leakage):**
- Churn Label — duplicate of Churn Value (text vs numeric), redundant
- Churn Score — likely an existing model's output, wouldn't exist at prediction time
- Churn Reason — only populated for already-churned customers (1,869 non-null 
  matches churned count), couldn't exist before the outcome
  
**Keep (legitimate feature):**
- CLTV — modest distributional difference by churn (means 4491 vs 4149, 
  overlapping ranges), unlike Churn Score/Reason; represents a value 
  calculable for any active customer, not derived from the churn outcome

**Target:**
- Churn Value (0/1) — keep, drop Churn Label

```python
df.groupby('Churn Value')['CLTV'].describe()
```

## Phase 1: Final Column Decisions

Drop:
- CustomerID, Count       (identifier / constant, no signal)
- Country, State           (constant, no signal)
- Lat Long, Latitude, Longitude, Zip Code   (redundant geography)
- Churn Label               (duplicate of Churn Value)
- Churn Score, Churn Reason  (target leakage)

Keep as feature:
- CLTV   (correlated but not derived from outcome)
- City   (deferred — will bucket top-N + "Other" in Phase 3 if EDA supports it)
- All service/account/demographic columns (Gender, Partner, Tenure Months, 
  Contract, Internet Service, etc.)

Target:
- Churn Value (0/1)

```python
cols_to_drop = ['CustomerID', 'Count', 'Country', 'State', 
                 'Lat Long', 'Latitude', 'Longitude', 'Zip Code',
                 'Churn Label', 'Churn Score', 'Churn Reason']

df_clean = df.drop(columns=cols_to_drop)
df_clean.shape
df_clean.columns.tolist()
```
